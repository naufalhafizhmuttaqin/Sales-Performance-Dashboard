**Data Cleaning**

```python
text_columns = df.select_dtypes(include="object").columns

df[text_columns] = df[text_columns].apply(
    lambda x: x.str.strip()
)
```

```python
df["Category"] = df["Category"].str.title()
```

```python
df["Segment"] = df["Segment"].str.title()
```

```python
df["State"] = df["State"].str.title()
```

```python
state_mapping = {
    "Califronia": "California",
    "Texsas": "Texas",
    "Newyork": "New York"
}

df["State"] = df["State"].replace(state_mapping)
```

```python
df.replace(
    "",
    np.nan,
    inplace=True
)
```

```python
date_columns = [
    "Order Date",
    "Ship Date"
]

for column in date_columns:

    df[column] = pd.to_datetime(
        df[column],
        format="mixed",
        errors="coerce"
    )
```

```python
df[date_columns].isna().sum()
```

```python
df.dropna(
    subset=date_columns,
    inplace=True
)
```

```python
df.drop_duplicates(
    inplace=True
)

df.reset_index(
    drop=True,
    inplace=True
)
```

```python
df["Customer Name"] = (
    df["Customer Name"]
      .fillna("Unknown Customer")
)
```

```python
df["Postal Code"] = (
    df["Postal Code"]
      .fillna(0)
      .astype(int)
)
```

```python
mode_region = df["Region"].mode()[0]

df["Region"] = (
    df["Region"]
      .fillna(mode_region)
)
```

```python
df["Product Name"] = (
    df["Product Name"]
      .fillna("Unknown Product")
)
```

```python
df = df[df["Sales"] > 0]
```

```python
df = df[df["Quantity"] > 0]
```

```python
df.loc[
    df["Discount"] > 1,
    "Discount"
] = 1
```

```python
df = df[
    df["Profit"] <= df["Sales"] * 2
]
```

```python
Q1 = df["Sales"].quantile(.25)
Q3 = df["Sales"].quantile(.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df = df[
    (df["Sales"] >= lower) &
    (df["Sales"] <= upper)
]
```

**Feature Engineering**

```python
df["Order Year"] = df["Order Date"].dt.year

df["Order Quarter"] = df["Order Date"].dt.quarter

df["Order Month"] = df["Order Date"].dt.month

df["Order Month Name"] = df["Order Date"].dt.month_name()

df["Order Weekday"] = df["Order Date"].dt.day_name()

df["Shipping Days"] = (
    df["Ship Date"] -
    df["Order Date"]
).dt.days

df["Profit Margin (%)"] = np.where(
    df["Sales"] != 0,
    (df["Profit"] / df["Sales"]) * 100,
    0
).round(2)

df["Sales per Unit"] = (
    df["Sales"] /
    df["Quantity"]
).round(2)

df["Profit per Unit"] = (
    df["Profit"] /
    df["Quantity"]
).round(2)
```
